## The Newton fractal 


Add theory from the rootfinding from the teaching


The newton fractal is obtained when one applies the Newton-Rhapson method for 
root finding, to a function of complex numbers. The fractal is generated by 
visualizing which roots the a set of numbers on the complex plane converge to.

$$x_{n+1} = x_{n} - \frac{f(x_{n})}{f'(x_{n})}$$

```python
import jax.numpy as jnp
from jax import grad
```

```python
f= lambda z: 1/(jnp.exp(z)-1)
```

```python
def newton(z0,f,maxit=1_000,tol=1e-6):
    fprime=grad(f)
    z=z0
    for i in maxit:
        dz=f(z)/fprime(z)
        if abs(z)<tol:
            return z
        z-=dz
    print("Convergence not reached")
    return False
```
